#Ejercicio 1
Simula 10 000 lanzamientos de dos dados y estima la probabilidad de que la suma sea exactamente 5. Compara esa frecuencia con la probabilidad clásica que ya calculaste.

```python
import random
def lanzar_dado():
    return random.randint(1, 6)

n = 10000
conteo_suma_5 = 0

for i in range(n):
    dado1 = lanzar_dado()
    dado2 = lanzar_dado()
    if dado1 + dado2 == 5:
        conteo_suma_5 += 1

prob_frecuentista = conteo_suma_5 / n
prob_clasica = 4 / 36

print(f"Probabilidad Frecuentista: {prob_frecuentista:.4f}")
print(f"Probabilidad Clásica:     {prob_clasica:.4f}")
```

#Ejercicio 3
Simula 20 000 lanzamientos de tres dados. Estima la frecuencia de la suma 9 y la de la suma 10. ¿Se nota, en la simulación, cuál de las dos sale más?

```python
def lanzar_dado():
    return random.randint(1, 6)

n = 20000
conteo_suma_9 = 0
conteo_suma_10 = 0

for i in range(n):
    d1, d2, d3 = lanzar_dado(), lanzar_dado(), lanzar_dado()
    suma = d1 + d2 + d3

    if suma == 9:
        conteo_suma_9 += 1
    elif suma == 10:
        conteo_suma_10 += 1

prob_frec_9 = conteo_suma_9 / n

print(f"\nSuma 9: {prob_frec_9:.4f}")
print(f"Suma 10: {prob_frec_10:.4f}")

if conteo_suma_10 > conteo_suma_9:
    print(f"\nRespuesta: En la simulación salió más la suma 10 por {conteo_suma_10 - conteo_suma_9} apariciones")
else:
    print(f"\nRespuesta: En la simulación salió más la suma 9 por {conteo_suma_9 - conteo_suma_10} apariciones")
```

#Ejercicio 4
Basado en el Ejemplo 3, simula una urna con 3 bolas rojas y 2 azules. Realiza 2,000 extracciones (con reemplazo) y calcula la probabilidad de obtener una azul.

```python
urna = ["roja", "roja", "roja", "azul", "azul"]

n = 2000
conteo_azules = 0

for i in range(n):
    if random.choice(urna) == "azul":
        conteo_azules += 1
prob_frecuentista = conteo_azules / n
print(f"Probabilidad Frecuentista:{prob_frecuentista:.4f}")
```

#Ejercicio 5
Una serie son 20 volados. Estima, con 10 000 series, la probabilidad de que aparezcan 5 águilas seguidas en algún momento de la serie.

```python
def lanzar_moneda():
    return random.choice(["águila", "sol"])

n = 10000
series_favorables = 0

for i in range(n):
    serie = [lanzar_moneda() for _ in range(20)]

    racha_actual = 0
    racha_maxima = 0
    for resultado in serie:
        if resultado == "águila":
            racha_actual += 1
            if racha_actual > racha_maxima:
                racha_maxima = racha_actual
        else:
            racha_actual = 0

    if racha_maxima >= 5:
        series_favorables += 1

prob_frecuentista = series_favorables / n

print(f"Probabilidad Frecuentista: {prob_frecuentista:.4f}")
```

#Ejercicio 6 — Bonus
Los clientes llegan a una tienda de forma aleatoria (entre 1 y 5 minutos). El cajero tarda entre 2 y 4 minutos en atender. Simula la llegada de 500 clientes y calcula la probabilidad de que un cliente tenga que esperar más de 10 minutos en la fila.

```python
n = 500
clientes_esperan_mas_10min = 0

tiempo_actual = 0
tiempo_cajero_libre = 0

for i in range(n):
    llegada = random.uniform(1, 5)
    tiempo_actual += llegada

    if tiempo_actual < tiempo_cajero_libre:
        espera = tiempo_cajero_libre - tiempo_actual
    else:
        espera = 0

    if espera > 10:
        clientes_esperan_mas_10min += 1
    atencion = random.uniform(2, 4)

    inicio = max(tiempo_actual, tiempo_cajero_libre)
    tiempo_cajero_libre = inicio + atencion

prob_frecuentista = clientes_esperan_mas_10min / n

print("Clientes que esperaron más de 10 min:", clientes_esperan_mas_10min)
print(f"Probabilidad Frecuentista: {prob_frecuentista:.4f}")
```
