#Ejercicio 2
Se lanzan tres dados. Calcula la probabilidad de que la suma sea 9, dado que las tres caras son distintas.

```python
from itertools import product
S = list(product(range(1, 7), repeat=3))

distintos = [t for t in S if len(set(t)) == 3]
suma_9 = [t for t in distintos if sum(t) == 9]

print(f"Probabilidad: {len(suma_9)} / {len(distintos)}")
```

#Teorema de Probabilidad Total

#Ejercicio 1
El 70% de los clientes ven un anuncio y, de ellos, compra el 30%. El 30% no ve el anuncio y, de ellos, compra el 10%. Calcula la probabilidad total de que un cliente compre.

```python
P_Anuncio = 0.70
P_NoAnuncio = 0.30

P_Compra_dado_Anuncio = 0.30
P_Compra_dado_NoAnuncio = 0.10

P_Compra = (P_Anuncio * P_Compra_dado_Anuncio) + (P_NoAnuncio * P_Compra_dado_NoAnuncio)

print(f"Probabilidad total de compra: {P_Compra:.2f}")
```

#Ejercicio 2
Se sabe que:

De esos, el 75% llueve

El 60% de los días no son nublados

De esos, el 10% llueve

¿Cuál es la probabilidad total de que llueva?

```python
P_Nublado = 0.40
P_NoNublado = 0.60

P_Lluvia_dado_Nublado = 0.75
P_Lluvia_dado_NoNublado = 0.10

P_Lluvia = (P_Nublado * P_Lluvia_dado_Nublado) + (P_NoNublado * P_Lluvia_dado_NoNublado)

print(f"Probabilidad total de lluvia: {P_Lluvia:.2f}")
```

#Ejercicio 3
Hay dos cajas de audífonos. La primera tiene 2 que sirven y 6 que fallan. La segunda tiene 5 que sirven y 3 que fallan. Se elige una caja al azar y después un audífono. ¿Cuál es la probabilidad de que sirva?

```python
P_Caja1 = 0.50
P_Caja2 = 0.50

P_Sirve_dado_Caja1 = 2 / 8

P_Sirve_dado_Caja2 = 5 / 8

P_Sirve = (P_Caja1 * P_Sirve_dado_Caja1) + (P_Caja2 * P_Sirve_dado_Caja2)

print(f"Probabilidad total de que sirva: {P_Sirve:.4f}")
```

#Teorema de Bayes

#Ejercicio 1
El 20% de los correos son spam. La palabra «oferta» aparece en el 80% de los correos spam y en el 10% de los correos deseados. Si un correo tiene la palabra «oferta», ¿cuál es la probabilidad de que sea spam?

```python
P_Spam = 0.20
P_Deseado = 0.80

P_Oferta_dado_Spam = 0.80
P_Oferta_dado_Deseado = 0.10

P_Oferta = (P_Spam * P_Oferta_dado_Spam) + (P_Deseado * P_Oferta_dado_Deseado)

P_Spam_dado_Oferta = (P_Spam * P_Oferta_dado_Spam) / P_Oferta

print(f"Probabilidad de que sea Spam dado que tiene 'oferta': {P_Spam_dado_Oferta:.4f}")
```

#Ejercicio 2
En una tienda online, el 40% entra por Chrome, el 35% por Safari y el 25% por otros. La tasa de conversión (compra) es del 3% en Chrome, 4% en Safari y 2% en otros.

Si se registró una venta, ¿qué navegador es más probable que haya usado el cliente?

```python
P_Chrome = 0.40
P_Safari = 0.35
P_Otros = 0.25

P_Venta_dado_Chrome = 0.03
P_Venta_dado_Safari = 0.04
P_Venta_dado_Otros = 0.02

P_Venta = (P_Chrome * P_Venta_dado_Chrome) + (P_Safari * P_Venta_dado_Safari) + (P_Otros * P_Venta_dado_Otros)

P_Chrome_dado_Venta = (P_Chrome * P_Venta_dado_Chrome) / P_Venta
P_Safari_dado_Venta = (P_Safari * P_Venta_dado_Safari) / P_Venta
P_Otros_dado_Venta = (P_Otros * P_Venta_dado_Otros) / P_Venta

print(f"Probabilidad de Chrome dado Venta: {P_Chrome_dado_Venta:.4f}")
print(f"Probabilidad de Safari dado Venta: {P_Safari_dado_Venta:.4f}")
print(f"Probabilidad de Otros dado Venta:  {P_Otros_dado_Venta:.4f}")

print("Respuesta: Es más probable que el cliente haya usado Safari.")
```
